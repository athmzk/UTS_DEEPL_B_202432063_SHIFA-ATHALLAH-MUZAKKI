  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432063_Shifa Athallah Muzakki
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim  # Modul optimizer (Adam, SGD, dll.)
from torch.utils.data import TensorDataset, DataLoader  # Membungkus tensor menjadi dataset dan batch
import re  # Regular expression untuk membersihkan teks
from collections import Counter  # Menghitung frekuensi kata untuk membuat vocabulary
from sklearn.model_selection import train_test_split  # Membagi data latih dan data uji
from sklearn.metrics import classification_report, confusion_matrix  # Metrik evaluasi klasifikasi

torch.manual_seed(42)  # Mengunci seed agar hasil bisa direproduksi
np.random.seed(42)  # Mengunci seed NumPy dengan alasan yang sama
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432063_Shifa Athallah Muzakki
# 1. Muat dataset
DATA_PATH = 'pashto_sentiment_dataset.csv'
df = pd.read_csv(DATA_PATH)
print('Jumlah baris dan kolom:', df.shape)
print(df.dtypes)
df.head()

# 2. Tentukan kolom label dan kolom teks secara otomatis
label_col = min(df.columns, key=lambda c: df[c].nunique())
text_cols = [c for c in df.columns if not pd.api.types.is_numeric_dtype(df[c]) and c != label_col]
text_col = max(text_cols, key=lambda c: df[c].astype(str).str.len().mean())
print('Kolom teks :', text_col, '| Kolom label :', label_col)

n_awal = len(df)
df = df[[text_col, label_col]].dropna().drop_duplicates().reset_index(drop=True)
print('Baris awal:', n_awal, '| setelah dibersihkan:', len(df))

# 3. Encoding label menjadi angka 0..K-1
label_names = sorted(df[label_col].unique(), key=str)
label2id = {lab: i for i, lab in enumerate(label_names)}
df['label_id'] = df[label_col].map(label2id)
num_classes = len(label_names)
print('Kelas:', label2id)
print(df['label_id'].value_counts())

# 4. Pembersihan teks dan tokenisasi
def clean_tokenize(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|@\w+|#\w+', ' ', text)
    text = re.sub(r'[^\w\s]', ' ', text)
    text = re.sub(r'\d+', ' ', text)
    return text.split()

df['tokens'] = df[text_col].apply(clean_tokenize)
df = df[df['tokens'].str.len() > 0].reset_index(drop=True)

# 5. Bagi data latih (80%) dan data uji (20%) sebelum membuat vocabulary
train_df, test_df = train_test_split(
    df, test_size=0.2, random_state=42, stratify=df['label_id'])

# 6. Vocabulary hanya dari data latih (mencegah kebocoran data)
MAX_VOCAB = 10000
MAX_LEN = 50
counter = Counter(w for toks in train_df['tokens'] for w in toks)
vocab = {'<PAD>': 0, '<UNK>': 1}
for w, _ in counter.most_common(MAX_VOCAB - 2):
    vocab[w] = len(vocab)
print('Ukuran vocabulary:', len(vocab))

# 7. Ubah kata menjadi indeks dan samakan panjangnya (padding)
def encode(tokens):
    ids = [vocab.get(w, 1) for w in tokens][:MAX_LEN]
    return [0] * (MAX_LEN - len(ids)) + ids

# 8. Konversi ke PyTorch Tensor
X_train = torch.tensor([encode(t) for t in train_df['tokens']], dtype=torch.long)
y_train = torch.tensor(train_df['label_id'].values, dtype=torch.long)
X_test = torch.tensor([encode(t) for t in test_df['tokens']], dtype=torch.long)
y_test = torch.tensor(test_df['label_id'].values, dtype=torch.long)
print('X_train:', X_train.shape, '| X_test:', X_test.shape)

# Diacak
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=64, shuffle=True)
# Tidak diacak
test_loader = DataLoader(TensorDataset(X_test, y_test), batch_size=64, shuffle=False)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432063_Shifa Athallah Muzakki
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_size, num_classes):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.rnn = nn.RNN(embed_dim, hidden_size, batch_first=True)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        emb = self.embedding(x)
        out, h_n = self.rnn(emb)
        last = h_n[-1]
        return self.fc(self.dropout(last))
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432063_Shifa Athallah Muzakki

# 1. Hyperparameter
epochs = 15
learning_rate = 0.001
embed_dim = 64
hidden_size = 64

# 2. Model, loss, dan optimizer
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = SequenceModel(len(vocab), embed_dim, hidden_size, num_classes).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. Riwayat untuk visualisasi
train_losses, test_losses = [], []
train_accs, test_accs = [], []

# 4. Training loop
for epoch in range(epochs):
    model.train()
    run_loss, benar, total = 0.0, 0, 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        logits = model(xb)
        loss = criterion(logits, yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        run_loss += loss.item() * xb.size(0)
        benar += (logits.argmax(1) == yb).sum().item()
        total += xb.size(0)
    train_losses.append(run_loss / total)
    train_accs.append(benar / total)

    model.eval()
    t_loss, t_benar, t_total = 0.0, 0, 0
    with torch.no_grad():
        for xb, yb in test_loader:
            xb, yb = xb.to(device), yb.to(device)
            logits = model(xb)
            t_loss += criterion(logits, yb).item() * xb.size(0)
            t_benar += (logits.argmax(1) == yb).sum().item()
            t_total += xb.size(0)
    test_losses.append(t_loss / t_total)
    test_accs.append(t_benar / t_total)
    print(f'Epoch {epoch+1:02d}/{epochs} | loss {train_losses[-1]:.4f} | acc {train_accs[-1]:.4f} | val_loss {test_losses[-1]:.4f} | val_acc {test_accs[-1]:.4f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432063_Shifa Athallah Muzakki

# 1. Grafik riwayat pelatihan (Loss dan Accuracy terhadap Epoch)
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
ax[0].plot(test_losses, color='#1FA4B1', linewidth=2, label='Test Loss')
ax[0].set_title('Grafik Penurunan Loss Selama Pelatihan')
ax[0].set_xlabel('Epoch')
ax[0].set_ylabel('Loss')
ax[0].grid(True, linestyle='--', alpha=0.5)
ax[0].legend()
ax[1].plot(train_accs, color='#114F6A', linewidth=2, label='Training Accuracy')
ax[1].plot(test_accs, color='#1FA4B1', linewidth=2, label='Test Accuracy')
ax[1].set_title('Grafik Akurasi Selama Pelatihan')
ax[1].set_xlabel('Epoch')
ax[1].set_ylabel('Accuracy')
ax[1].grid(True, linestyle='--', alpha=0.5)
ax[1].legend()
plt.tight_layout()
plt.show()

# 2. Prediksi pada data uji
model.eval()  # Mode evaluasi
all_pred, all_true, all_conf = [], [], []
with torch.no_grad():
    for xb, yb in test_loader:
        probs = torch.softmax(model(xb.to(device)), dim=1)
        conf, pred = probs.max(dim=1)
        all_pred.extend(pred.cpu().numpy())
        all_true.extend(yb.numpy())
        all_conf.extend(conf.cpu().numpy())
all_pred, all_true, all_conf = np.array(all_pred), np.array(all_true), np.array(all_conf)

# 3. Distribusi hasil prediksi sentimen dan skornya
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
x = np.arange(num_classes)
ax[0].bar(x - 0.2, np.bincount(all_true, minlength=num_classes), width=0.4, color='#114F6A', label='Data Aktual')
ax[0].bar(x + 0.2, np.bincount(all_pred, minlength=num_classes), width=0.4, color='#1FA4B1', label='Prediksi Model')
ax[0].set_xticks(x)
ax[0].set_xticklabels([str(l) for l in label_names])
ax[0].set_title('Distribusi Sentimen: Aktual vs Prediksi')
ax[0].set_ylabel('Jumlah Data Uji')
ax[0].legend()
ax[1].hist(all_conf, bins=20, color='#1FA4B1', edgecolor='#114F6A')
ax[1].set_title('Distribusi Skor Keyakinan Prediksi')
ax[1].set_xlabel('Skor (probabilitas kelas terprediksi)')
ax[1].set_ylabel('Jumlah Data')
plt.tight_layout()
plt.show()

# 4. Metrik evaluasi
print('Akurasi data uji:', (all_pred == all_true).mean())
print(classification_report(all_true, all_pred, target_names=[str(l) for l in label_names]))
print('Confusion matrix:\n', confusion_matrix(all_true, all_pred))
```
